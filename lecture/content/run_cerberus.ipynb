## User input data Submission for running Cerberus base parameters
In the modular run instead of a sample sheet, the input information will be provided directly to the GeneticInteractionSample class. All variables that are initialized when instantiating the class can or have to be overwritten. Mandatory parameter have to been set. Overview on the mandatory base paramters:

| parameter | explanation |
| --------- | ----------------------------------------- |
| cell_line | name of the cell line, if RPE1 use 'RPE1' |
| reco_table_file | (absolute or relative) path to read count table |
| query_casette | guide column and gene column of the query casette |
| other_casette | guide column and gene column of the non-query casette |
| t0_tend_reco_mapping | {library/early timepoint 1: [end timepoint 1, endtime point 2], library/early timepoint 2: [...]}  most experiments have exactly one start point|
| dataset_name | human readable name for your dataset, can contain spaces |
| tend_reco_rep_pairs | pairwise list of lists of your end timepoints: e.g for 3 reps: [[rep1, rep2], [rep1, rep3], [rep2, rep3]] |
| neutral_control_gene | non targeting or save harbor targeting control gene name |
| conditions | at the moment placeholder set empty dictionray: {} |
| bio_reps | if any {bio1: pre/suffix 1, bio2: pre/suffix 2, ...} else: {} |
| tech_reps | dictionary similar to bio_reps but with tech rep pre/suffix |
| guide_reps | dictionary that shows guide id suffix, more details in chapter on gRNA id style guide |
| orientation_reps | True/False, set True if you have screened each pair in both directions and want to use it as replicate |
| guide_combination | dictionary of allowed guide combinations, will show below for a 4 guide library how to set up |
| output_folder | full path to the folder name to which results shall be written, output folder does not need to pre-exist |

Overview on the optional base parameters:
| parameter | default value | when and how to overwrite |
| --------- | ------------- | ------------------------- |
| merge_orientations | False | set to True if gRNA pairs were screened in both orientations (symmetric set up) and you assume high orientation agreement |
| genes_to_remove | [] | fill the list with gene names that shall be excluded from the analysis |
| custom_ess | [] | fill if you have an own list of essential genes for your screen and you dont want to use builtin list like Hart essentials or depmap essentials |
| custom_noness | [] | fill if you have an own list of nonessential genes for your screen and you dont want to use builtin lists |
| custom_enh_prolif | [] | fill if you have an own list of proliferation enhancing genes for your screen, if you not use RPE1 there will be no builtin set, and this set will be omitted from QC |
| guide_cov | 100 | synthetic guide coverage |
| add | 1 | fudge factor for normalizing read counts, this is added before doing logarithm |
| low_reco_cut_off | 10 | recommendation is to use the second percentile, if this is lower than 10 you will be asked to increase. Recommended is at least a read count of 20 as low read count cut-off|


## User input data Submission for running the multiplicative model module
In the modular run instead of a sample sheet, the input information will be provided directly to the GeneticInteractionSample class. All variables that are initialized when instantiating the class can or have to be overwritten. Mandatory parameter have to been set. Optional parameters for the multiplicative model are:

| parameter | explanation |
| --------- | ----------- |
| adjust_gis | set to True if you want to allow corrections to the rigid multiplicative model, False if you want the raw scores. Adjustments made: Ma transformation of raw scores, loess fit through raw scores, adjusting scores considering sd|
| sampling | only relevant, when adjusting, then adviced to set to True, to decrease runtime for fitting through the raw scores |
| scaling | select one of the three options: No Scaling / Nonessential Scaling / Ceres Scaling. Nonessential scaling: median of nonessential genes is set to 0.Ceres Scaling: median of nonessential genes set to 0, essential genes to -1. |
| n_sample | default is 10, adjust how many datapoints shall be sampled per bin |
| n_bins | default is 8, how many bins over the x axis you want to have |
| min_points | default is 30, minimum points per bin, that are necessary in order to conduct the fitting |
| frac | default 0.666, only change if you have very good reason to. This controls the neighborhood of points being considered for local fitting |
| degree | default is 1, some publications go with 2 (quadratic fit). Since we do local fitting, 1 is enough to account for non-linear trends |
| store_raw_gi_scores | default is False, set to True if you set adjusting_gis to True, but you also want the raw scores as output |
| lower_quant | default is 10, used for sd adjusting, everything lower than lower quantile will not be considered, so you protect the potential hits from the adjusting |
|upper_quant | default is 90, used for sd adjusting, everything higher than upper quantile will not be considered, so you protect the potential hits from the adjusting |
|opt_exclusion | default is 1, number of remaining guide combinations that need to remain after dropping low read count instances and loss of linearity instances to be considered |

# User Input to run gene level aggregation and statistics
The following parameters are required to execute limma. For most of them it is adviced to keep them set to their default values
| parameter | default value | explanation |
|-----------|---------------|-------------|
| scores_file | - | provide the file path to where the limma input table was written to. Cerberus can handle this internally|
| output_directory | - | proved the full path to where the limma output should be written to. Recommended to let Cerberus handle it internally |
| FDR | BH | FDR method, default is Benjamini Hochberg. Other methods can be chosen but are not recommended |
| skip_update | TRUE | do not pull the latest CeRberus code from github when running, set to FALSE if you want to update everytime |
| verbose | TRUE | get detailed output on which steps limma is running atm, set to FALSE, to omit progress being printed to screen |
| keep_all_configurations | FALSE | keep all results of other tested configurations of running limma, beside the "optimal" one |
